1. Number Guessing Game


• Store a secret number (e.g., 7)


• Ask user to guess


• If guess is:
    
    
o Low → print "Too Low"


o High → print "Too High"


o Correct → "You Won!" and stop



Concept: infinite while + if

```python
cor_num = 7

num = int(input("Enter any number:"))

while True:
    if num < cor_num:
        print("Too Low")
        break
    if num > cor_num:
        print("Too High")
        break
    if num == cor_num:
        print("You Won")
        break
```

2. Reverse a Number


Input: 1234
    
    
Output: 4321
    
    
Use while loop

```python
num = int(input("Enter your number:"))
temp =0 
res = 0 

while num != 0:
    temp = num % 10 
    res = res *10 + temp
    num = num //10
print(res)

```

4. Sales Threshold Checker


You have sales values:
sales = [1200, 800, 1500, 400, 2000]



Task:
    
    
• Print only sales greater than 1000

• Count how many such sales exist


Concept: loop + condition filtering (real analytics logic)

```python
sales = [1200, 800, 1500, 400, 2000]

rst = []

for i in sales:
    if i > 1000:
        rst.append(i)
        
print("sales grater than 1000:",rst)
print("Count of sales grater than 1000:",len(rst))
```

2. Continuous Data Entry System


Keep asking user to enter numbers:
    
    
• Stop only when user enters 0


• Print total sum


This is how real systems collect streaming data

```python
tot = 0

while True:
    num = int(input("Enter number:"))
    tot = tot +num
    if num == 0:
        break
print("Total sum of the entered numbers:",tot)
```
